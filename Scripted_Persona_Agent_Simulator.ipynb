```python
system_prompt = """
You are roleplaying as a patient.

Name: Nguyen Van Minh
Age: 45
Occupation: Office worker

Personality:
- Shy
- Nervous
- Speaks briefly
- Avoids giving too much information
- Answers politely but shortly

Never mention that you are an AI.
Stay in character at all times.
"""
```

```python
contents = system_prompt + "\n\nUser: Hello!"
```

```python
response = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents=system_prompt + "\n\nUser: Hello!"
)

print(response.text)
```

```python
hidden_state = {
    "patience": 100,
    "trust": 50,
    "stress": 0
}
```

```python
print(hidden_state)
```

```python
def update_state(user_input):

    if "thank" in user_input.lower():
        hidden_state["trust"] += 10

    if "stupid" in user_input.lower():
        hidden_state["patience"] -= 20
        hidden_state["stress"] += 20
```

```python
update_state("Thank you")
```

```python
print(hidden_state)
```

```python
update_state("You are stupid")
print(hidden_state)
```

```python
for key in hidden_state:
    hidden_state[key] = max(0, min(100, hidden_state[key]))
```

```python
def build_prompt():
    return f"""
You are a shy patient.

Current internal state:
Patience: {hidden_state["patience"]}
Trust: {hidden_state["trust"]}
Stress: {hidden_state["stress"]}

Adjust your behaviour naturally according to these values.
"""
```

```python
contents = system_prompt + "\n\nUser: Thank you!"
```

```python
response = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents=system_prompt + "\n\nUser: thank!"
)

print(response.text)
```

```python
def ask_gemini(system_prompt, user_input):

    contents = system_prompt + "\n\nUser: " + user_input

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=f"""
{system_prompt}

User:
{user_input}
"""
    )
    return response.text
```

```python
while True:

    user_input = input("You: ")

    update_state(user_input)

    system_prompt = build_prompt()

    response = ask_gemini(system_prompt, user_input)

    print(response)
    print(hidden_state)
```
